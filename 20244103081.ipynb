```python
# Here are the different types of data types in Python

x = "Hello World"
print(type(x))  # x is a str -> string


x1 = 20
print(type(x1))  # x1 is an int -> integer


x2 = 20.5
print(type(x2))  # x2 is a float -> floating-point number


x3 = 1j
print(type(x3))  # x3 is a complex -> complex number


x4 = ["apple", "banana", "cherry"]
print(type(x4))  # x4 is a list -> ordered and changeable collection


x5 = ("apple", "banana", "cherry")
print(type(x5))  # x5 is a tuple -> ordered and unchangeable collection


x6 = range(6)
print(type(x6))  # x6 is a range -> sequence of numbers


x7 = {"name": "John", "age": 36}
print(type(x7))  # x7 is a dict -> dictionary with key-value pairs


x8 = {"apple", "banana", "cherry"}
print(type(x8))  # x8 is a set -> unordered collection of unique values


x9 = frozenset({"apple", "banana", "cherry"})
print(type(x9))  # x9 is a frozenset -> unchangeable set


x10 = True
print(type(x10))  # x10 is a bool -> boolean value (True or False)


x11 = b"Hello"
print(type(x11))  # x11 is a bytes -> immutable sequence of bytes


x12 = bytearray(5)
print(type(x12))  # x12 is a bytearray -> changeable sequence of bytes


x13 = memoryview(bytes(5))
print(type(x13))  # x13 is a memoryview -> allows access to memory of another object


x14 = None
print(type(x14))  # x14 is NoneType -> represents no value
```

Assignment Operator

```python
# Assignment operators in Python

# =  -> Assigns a value to a variable
x = 5
print(x)  # Output: 5


# += -> Adds a value and assigns the result to the variable
x = 5
x += 3  # Same as: x = x + 3
print(x)  # Output: 8


# -= -> Subtracts a value and assigns the result
x = 5
x -= 3  # Same as: x = x - 3
print(x)  # Output: 2


# *= -> Multiplies a value and assigns the result
x = 5
x *= 3  # Same as: x = x * 3
print(x)  # Output: 15


# /= -> Divides a value and assigns the result
x = 5
x /= 3  # Same as: x = x / 3
print(x)  # Output: 1.6666666666666667


# %= -> Finds the remainder and assigns the result
x = 5
x %= 3  # Same as: x = x % 3
print(x)  # Output: 2


# //= -> Performs floor division and assigns the result
x = 5
x //= 3  # Same as: x = x // 3
print(x)  # Output: 1


# **= -> Raises to a power and assigns the result
x = 5
x **= 3  # Same as: x = x ** 3
print(x)  # Output: 125


# &= -> Performs bitwise AND and assigns the result
x = 5
x &= 3  # Same as: x = x & 3
print(x)  # Output: 1


# |= -> Performs bitwise OR and assigns the result
x = 5
x |= 3  # Same as: x = x | 3
print(x)  # Output: 7


# ^= -> Performs bitwise XOR and assigns the result
x = 5
x ^= 3  # Same as: x = x ^ 3
print(x)  # Output: 6


# >>= -> Shifts the bits to the right and assigns the result
x = 5
x >>= 3  # Same as: x = x >> 3
print(x)  # Output: 0


# <<= -> Shifts the bits to the left and assigns the result
x = 5
x <<= 3  # Same as: x = x << 3
print(x)  # Output: 40


# := -> Assignment expression (walrus operator)
# It assigns a value and uses that value in the same expression
print(x := 3)  # Output: 3

# Same as:
# x = 3
# print(x)
```

```python
thistuple = ("apple", "banana", "cherry")
print(thistuple) #('apple', 'banana', 'cherry')
#a tuples can also be created without parentheses
thistuple = "apple", "banana", "cherry"
print(thistuple) #('apple', 'banana', 'cherry')
thistuple = ("apple", "banana", "cherry", "apple", "cherry")
print(thistuple)
#('apple', 'banana', 'cherry', 'apple', 'cherry')
```

```python
#length of a tuple can be find using len()
thistuple = ("apple", "banana", "cherry")
print(len(thistuple)) #3
#tuple with one item need to give a , after the item unless it will be a string
thistuple = ("apple",)
print(type(thistuple))

#NOT a tuple
thistuple = ("apple")
print(type(thistuple))
#empty tuple
thistuple = ()
print(type(thistuple))
#tuple can contains any data type
tuple1 = ("abc", 34, True, 40, "male")
print(tuple1) #('abc', 34, True, 40, 'male')
#It is also possible to use the tuple() constructor to make a tuple.
thistuple = tuple(("apple", "banana", "cherry")) # note the double round-brackets
print(thistuple)
```

```python
#Access Tuple item like string and list
thistuple = ("apple", "banana", "cherry")
print(thistuple[1]) #banana
#negative indexing
thistuple = ("apple", "banana", "cherry")
print(thistuple[-1]) #cherry
#Return the third, fourth, and fifth item:
thistuple = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")
print(thistuple[2:5])
```

```python
#Intersection-> intersection() method will return a new set, that only contains the items that are present in both sets
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1.intersection(set2)
print(set3) #apple , mainly returns the common item between them
```

```python
#difference()
#The difference() method will return a new set that will contain only the items from the first set that are not present in the other set.
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1.difference(set2)

print(set3) #{'banana', 'cherry'}
#can be use - operator but need a extra variable
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1 - set2
print(set3) #{'banana', 'cherry'}
#in the method difference_update() do not need a extra variable
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set1.difference_update(set2)

print(set1)  #{'banana', 'cherry'}


```

```python
#The symmetric_difference() method will keep only the elements that are NOT present in both sets.
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1.symmetric_difference(set2)

print(set3)  #only print the items that's are not present in both
#{'microsoft', 'banana', 'google', 'cherry'}
#also can use as ^
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1 ^ set2
print(set3)
```

```python
#Python frozenset
#frozenset is an immutable version of a set.

#Like sets, it contains unique, unordered, unchangeable elements.

#Unlike sets, elements cannot be added or removed from a frozenset.
x = frozenset({"apple", "banana", "cherry"})
print(x)
print(type(x))
#some operations of frozenset
# Frozenset Operations

set1 = frozenset(["apple", "banana", "cherry"])
set2 = frozenset(["banana", "orange", "mango"])

# 1. copy()
print("copy():", set1.copy())

# 2. difference()
print("difference():", set1.difference(set2))
print("difference (-):", set1 - set2)

# 3. intersection()
print("intersection():", set1.intersection(set2))
print("intersection (&):", set1 & set2)

# 4. isdisjoint()
print("isdisjoint():", set1.isdisjoint(set2))

# 5. issubset()
set3 = frozenset(["apple", "banana"])
print("issubset():", set3.issubset(set1))
print("issubset (<=):", set3 <= set1)

# 6. issuperset()
print("issuperset():", set1.issuperset(set3))
print("issuperset (>=):", set1 >= set3)

# 7. symmetric_difference()
print("symmetric_difference():", set1.symmetric_difference(set2))
print("symmetric_difference (^):", set1 ^ set2)

# 8. union()
print("union():", set1.union(set2))
print("union (|):", set1 | set2)
```
