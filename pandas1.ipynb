**pandas series**
a pandas series is like a column in a table.It's a 1D array holding data type

# This is formatted as code

````python
import numpy as np
import pandas as pd
````

series from list

````python
#string
country =['india','pakistan','bangladesh','srilanka']
pd.Series(country)
````

````python
#integer
runs =[12,55,81,100,10]
runs_ser =pd.Series(runs)
runs_ser
````

````python
#custom index
marks = [67,57,89,100]
subjects =['math','english','sciecne','bangla']
pd.Series(marks,index = subjects)
````

````python
#setting a name
marks = pd.Series(marks,index=subjects,name='marks of bro')
marks
````

series from dict

````python
marks = {
    'math':67,
    'english':57,
    'science':89,
    'bangla':100
}
marks_series = pd.Series(marks,name='marks of man')
marks_series
````
